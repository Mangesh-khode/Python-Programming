```python
list1=[1, 2, 3, 2, 4, 5, 1, 6, 3]
list2=[]
list3=[]

for i in list1:
    if i not in list2:
        list2.append(i)
        
    else:
        list3.append(i)
        
print(list2)
print(list3)
```

```python
numbers = [12, 7, 15, 4, 9, 2, 5]

smallest = None

for num in numbers:
    if num % 2 != 0:      # Check if number is odd
        if smallest is None or num < smallest:
            smallest = num

if smallest is not None:
    print("Smallest odd number:", smallest)
else:
    print("No odd number found.")
    
```
