```python
# 1. Print all multiples of 7 from 1 to 100.
# Solution:
for i in range(1, 101):
    if i % 7 == 0:
        print(i)
```

```python
# 2. Count the number of positive and negative numbers in a list.
# Solution:
numbers = [10, -5, 20, -8, 0, 15, -3]

positive = 0
negative = 0

for i in numbers:
    if i > 0:
        positive += 1
    elif i < 0:
        negative += 1

print("Positive numbers:", positive)
print("Negative numbers:", negative)
```

```python
# 4. Find the sum of numbers divisible by 3 between 1 and 100.
# Solution:
sum = 0

for i in range(1, 101):
    if i % 3 == 0:
        sum += i

print("Sum =", sum)
```

```python
# 5. Print all numbers between 1 and 100 that are divisible by 2 but not by 5.
# Solution:
for i in range(1, 101):
    if i % 2 == 0 and i % 5 != 0:
        print(i)
```

```python
# 6. Write a Python program to find the largest even number in a list.

numbers = [12, 45, 68, 23, 90, 17]

largest = 0

for i in numbers:
    if i % 2 == 0:
        if i > largest:
            largest = i

print("Largest even number is:", largest)
```

```python
# 7. Write a Python program to find the smallest odd number in a list.
numbers = [12, 45, 68, 23, 91, 17]

smallest = 91 #take large number to compare from list. i suggest use float("inf") mean most largest number

for i in numbers:
    if i % 2 != 0 and i < smallest:
        smallest = i

print("Smallest odd number is:", smallest)
```

```python
# 8. Count how many elements are divisible by both 2 and 3 in a list.
# Solution:

numbers = [6, 12, 15, 18, 20, 24, 30]

count = 0

for i in numbers:
    if i % 2 == 0 and i % 3 == 0:
        count += 1

print("Count =", count)
```

```python
# 1. A student has marks [85, 90, 78, 88, 92]. Find the total marks.
# Solution:

marks = [85, 90, 78, 88, 92]

total = 0

for i in marks:
    total += i

print("Total Marks =", total)
```
