```python
# Count Uppercase, Lowercase, Digits, Special Characters "Python123@AI"
text="Python123@AI"

upper=lower=digit=special=0

for ch in text:
    if ch.isupper():
        upper+=1
    elif ch.islower():
        lower+=1
    elif ch.isdigit():
        digit+=1
    else:
        special+=1

print(upper,lower,digit,special)
```

```python
# Find Duplicate Emails
emails=[
"a@gmail.com",
"b@gmail.com",
"a@gmail.com",
"c@gmail.com",
"b@gmail.com"
]

seen=set()

duplicates=set()

for email in emails:
    if email in seen:
        duplicates.add(email)
    else:
        seen.add(email)

print(duplicates)
```

```python
# 3. Validate Password

# Conditions

# Minimum 8 characters
# One uppercase
# One lowercase
# One digit
# One special character

password = "Python@123"

upper = lower = digit = special = 0

for ch in password:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digit += 1
    else:
        special += 1

if len(password)>=8 and upper and lower and digit and special:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
# 7. Check Whether a Sentence is Pangram (Contains all 26 alphabets).

sentence="The quick brown fox jumps over the lazy dog"

letters=set()

for ch in sentence.lower():
    if ch.isalpha():
        letters.add(ch)

print(len(letters)==26)
```

```python
# Rotate a List by One Position numbers = [1,2,3,4,5]
numbers = [1,2,3,4,5]

last = numbers[-1]

for i in range(len(numbers)-1,0,-1):
    numbers[i] = numbers[i-1]

numbers[0] = last

print(numbers)
```

```python
# Find the Row with Maximum Sum
matrix = [
    [1,2,3],
    [10,20,30],
    [4,5,6]
]

maximum = 0
row = 0

for i in range(len(matrix)):

    total = sum(matrix[i])

    if total > maximum:
        maximum = total
        row = i

print(row)
```

```python
# Write a Python program to find the third largest unique element in a list without using sort().
numbers = [12, 45, 67, 89, 45, 90, 23]

first = second = third = float('-inf')

for num in numbers:

    if num > first:
        third = second
        second = first
        first = num

    elif num > second and num != first:
        third = second
        second = num

    elif num > third and num != first and num != second:
        third = num

print(third)
```

```python
# Count Even and Odd Digits in a Number
num = 54823916

even = odd = 0

while num > 0:

    digit = num % 10

    if digit % 2 == 0:
        even += 1
    else:
        odd += 1

    num //= 10

print("Even:", even)
print("Odd:", odd)
```

```python
# Reverse a Number Without Converting to String
num = 12345

reverse = 0

while num > 0:

    digit = num % 10

    reverse = reverse * 10 + digit

    num //= 10

print(reverse)
```
