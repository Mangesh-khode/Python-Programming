```python
# 1. Write a Python program to toggle the case of each character in a string
# without using the built-in swapcase() method.

# Example:
# PyThOn

# Output:
# pYtHoN

s = "PyThOn"
result = ""

for ch in s:
    if ch.islower():
        result += ch.upper()
    else:
        result += ch.lower()

print(result)
```

```python
# 2. Print Armstrong numbers from 1 to 1000.
for num in range(1, 1001):
    temp = num
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** 3
        temp //= 10

    if total == num:
        print(num)
```

```python
# Second largest element in the list
nums = [10, 20, 4, 45, 99]

first = second = float('-inf')

for n in nums:
    if n > first:
        second = first
        first = n
    elif n > second and n != first:
        second = n

print("Second largest element:", second)
```

```python
# Find the longest palindrome word in a sentence.
# Ex:sentence = "madam, level and malayalam are palindrome words in this sentence"

sentence = "madam, level and malayalam are palindrome words in this sentence"

words = sentence.split()
longest = ""

for word in words:
    rev = ""
    
    # manually reverse the word
    for ch in word:
        rev = ch + rev
    
    # check palindrome
    if word == rev:
        if len(word) > len(longest):
            longest = word

print("Longest palindrome word:", longest)
```

```python
# Find the length of the longest substring containing unique characters only.
s = "abcabcbb"

max_len = 0

for i in range(len(s)):
    temp = ""

    for j in range(i, len(s)):
        if s[j] not in temp:
            temp += s[j]
            
        else:
            break

    if len(temp) > max_len:
        max_len = len(temp)

print(max_len)
```
