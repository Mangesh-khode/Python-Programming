```python
# Q5. WAP that counts the total number of vowels in a given sentence.
# Given Input: "Loops are Fun!"
# vowels = "aeiouAEIOU"

string="Loops are Fun!"
vowels="aeiouAEIOU"
count=0
for i in string:
    if i in vowels:
        count+=1
        
print(count)
```

```python
# Q11. Reverse a Number
num = int(input("Enter Number: "))
# 1234
reverse = 0 #4321

while num > 0: #12

    digit = num % 10

    reverse = reverse * 10 + digit

    num = num // 10

print("Reversed Number:", reverse)
```
