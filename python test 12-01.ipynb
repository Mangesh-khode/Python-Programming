```python
# Q1. Rotate List by K Positions
# nums = [1, 2, 3, 4, 5]
# k = 2

# result = nums[-k:] + nums[:-k]
# print(result)

nums = [1, 2, 3, 4, 5]
k = 2

for i in range(k):
    last = nums.pop()      # Remove last element
    nums.insert(0, last)   # Insert at beginning

print(nums)
```

```python
# Q1. Rotate List by K Positions
nums = [1, 2, 3, 4, 5, 6,7]
k = 4

result = nums[-k:] + nums[:-k]
print(result)
```

```python
# Q2. Move All Zeros to End
nums = [1, 0, 2, 0, 3, 4, 0]

count = nums.count(0)

for i in range(count):
    nums.remove(0)
    nums.append(0)

print(nums)
```

```python
# Q3. Find All Duplicate Elements
nums = [1,2,3,2,4,5,1,6]

dup = []

for i in nums:
    if nums.count(i) > 1 and i not in dup:
        dup.append(i)

print(dup)
```

```python
# Q5. Find All Pairs of two lists with Given Sum

list1 = [1, 2, 3, 4]
list2 = [5, 6, 7, 8]
target = 9

for i in list1:
    for j in list2:
        if i + j == target:
            print((i, j))
```
