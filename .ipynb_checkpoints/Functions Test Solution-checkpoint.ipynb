```python
# Write a function that accepts three numbers and returns the largest number.

def find_max(a, b, c):
    return max(a, b, c)

print(find_max(10, 25, 15))
```

```python
# Write a function that checks whether a number is even or odd.

def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(7))
```

```python
# Write a function to calculate the factorial of a number.

def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact *= i

    return fact

print(factorial(5))
```

```python
# Write a function to check whether a number is prime.

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(is_prime(11))
```

```python
# Write a function to reverse a string.

def reverse_string(text):
    return text[::-1]

print(reverse_string("Python"))
```

```python
# Write a function to count vowels in a string.

def count_vowels(text):
    count = 0

    for char in text.lower():
        if char in "aeiou":
            count += 1

    return count

print(count_vowels("Programming"))
```

```python
# Write a function that returns the sum of all elements in a list.

def list_sum(numbers):
    return sum(numbers)

print(list_sum([10, 20, 30, 40]))
```

```python
# Write a function to check whether a string is a palindrome.

def is_palindrome(text):
    return text == text[::-1]

print(is_palindrome("madam"))
```

```python
# Write a function to print the first n Fibonacci numbers.

def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b

fibonacci(10)
```

```python
# Write a function to count the number of digits in a number.

def count_digits(num):
    return len(str(num))

print(count_digits(12345))
```

```python
# Write a function that returns the largest element in a list.

def largest_element(numbers):
    return max(numbers)

print(largest_element([5, 12, 8, 20, 3]))
```

```python
# Write a function to check whether a number is an Armstrong number.

def is_armstrong(num):
    total = 0
    digits = len(str(num))

    for digit in str(num):
        total += int(digit) ** digits

    return total == num

print(is_armstrong(153))
```

```python
# Write a function to remove duplicate elements from a list.

def remove_duplicates(numbers):
    return list(set(numbers))

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))
```

```python
# Write a function that returns the second largest number in a list.

def second_largest(numbers):
    numbers = list(set(numbers))
    numbers.sort()
    return numbers[-2]

print(second_largest([10, 20, 30, 40, 50]))
```
