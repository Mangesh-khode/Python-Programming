```python
# 1. Find all pairs in a list whose sum is equal to a given target value
numbers = [2, 4, 3, 5, 7, 8, 9]
target = 10

for i in range(len(numbers)):#0,1,2,3,4,5,6
    for j in range(i + 1, len(numbers)):
        if numbers[i] + numbers[j] == target:
            print(numbers[i], numbers[j])
```

```python
# 2. Find all duplicate elements in a list
numbers = [1, 2, 3, 2, 4, 5, 1, 6, 3]

duplicates = []

for i in numbers:
    if numbers.count(i) > 1 and i not in duplicates:
        duplicates.append(i)

print("Duplicate elements:", duplicates)
```

```python
list1=[1, 2, 3, 2, 4, 5, 1, 6, 3]
list2=[]
list3=[]
for i in list1:
    if i not in list2:
        list2.append(i)
        
    else:
        list3.append(i)
        
print(list2)
print(list3)
```

```python
# 3. Count the number of even and odd elements in a tuple
numbers = (10, 15, 22, 33, 44, 51)

even = 0
odd = 0

for i in numbers:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
```

```python
# 4. Generate the Fibonacci series up to n terms
n = int(input("Enter number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
# 5. Reverse a given number
number = 1234

reverse = 0#4

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

print("Reversed Number:", reverse)
```

```python
# 6. Find the maximum and minimum elements in a set
numbers = {15, 40, 10, 80, 25}

print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
```

```python
# 7. Remove all digits from a given string
text = "Python123 Programming456"

result = ""

for ch in text:
    if not ch.isdigit():
        result += ch

print(result)
```

```python
# 8. Reverse a sentence
# Input
# python is an easy language

sentence = "python is an easy language"

words = sentence.split()
print(words)
reverse_words = words[::-1]

result = " ".join(reverse_words)

print(result)
```

```python
# 9. Count the frequency of each character in a string
text = "programming"

frequency = {}

for ch in text:
    frequency[ch] = frequency.get(ch, 0) + 1

print(frequency)
```

```python
# 10. List Comprehension for cube of each element in a list
numbers = [1, 2, 3, 4, 5]

cube = [i ** 3 for i in numbers]

print(cube)
```
