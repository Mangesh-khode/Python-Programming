# Nested If-Else Practice Questions

# 1. Question

A country has the following voting rules:

    1. A person must be 18 years or older to vote.
    2. If the person is 18 or older, then check whether the person is a citizen or not.
    3. Only citizens are allowed to vote.

Write a Python program using nested if-else to:

    1. Take age as input
    2. Ask whether the person is a citizen (yes/no)
    3. Display:
        "You can vote." if both conditions are true
        "You are not a citizen." if age is valid but citizenship is "no"
        "You are under 18." if age is less than 18

```python
# Taking age input from user
age = int(input("Enter your age: "))

# Taking citizenship input from user
citizen = input("Are you a citizen? (yes/no): ")

# First condition:
# Check if person's age is 18 or more
if age >= 18:
    
    # Second condition inside first if:
    # Check if person is a citizen
    if citizen == "yes" or "Yes":
        print("You can vote.")
    
    # If person is not a citizen
    else:
        print("You are not a citizen.")

# If age is less than 18
else:
    print("You are under 18.")
```

# 2. Question 

A school has the following rules to decide whether a student passes or not:

    1. A student must score 35 or more marks to pass the exam.
    2. If the student has passed the exam, then check the attendance percentage.
    3. The student must have 75% or more attendance to be considered fully eligible.
    4. If attendance is below 75%, display "Low attendance".

Write a Python program using nested if-else to:

    1. Take marks as input
    2. Take attendance percentage as input
    3. Display:
        "Student Passed" if marks are 35 or more and attendance is 75% or more
        "Low attendance" if marks are enough but attendance is less than 75%
        "Student Failed" if marks are below 35

```python
# Taking marks input from user
marks = int(input("Enter marks: "))

# Taking attendance percentage input
attendance = int(input("Enter attendance percentage: "))

# First condition:
# Check if student passed the exam
if marks >= 35:
    
    # Second condition inside first if:
    # Check attendance percentage
    if attendance >= 75:
        print("Student Passed")
    
    # If attendance is less than 75
    else:
        print("Low attendance")

# If marks are less than 35
else:
    print("Student Failed")
```

# 3. Question

An ATM machine has the following withdrawal rules:

    1. The user must enter the correct PIN to access the account.
    2. If the PIN is correct, then check the withdrawal amount.
    3. The withdrawal amount must be less than or equal to the available balance.
    4. If the balance is enough, display "Withdrawal Successful".
    5. If the balance is not enough, display "Insufficient Balance".
    6. If the PIN is incorrect, display "Wrong PIN".

Write a Python program using nested if-else to:

    1. Take the PIN as input
    2. Take the withdrawal amount as input only if the PIN is correct
    3. Display:
        "Withdrawal Successful" if both conditions are true
        "Insufficient Balance" if balance is low
        "Wrong PIN" if the entered PIN is incorrect

```python
# Original ATM PIN
pin = 1234

# Taking PIN input from user
user_pin = int(input("Enter PIN: "))

# First condition:
# Check whether entered PIN is correct
if user_pin == pin:
    
    # Account balance
    balance = 5000
    
    # Taking withdrawal amount input
    amount = int(input("Enter withdrawal amount: "))
    
    # Second condition inside first if:
    # Check if balance is enough
    if amount <= balance:
        print("Withdrawal Successful")
    
    # If balance is not enough
    else:
        print("Insufficient Balance")

# If PIN is incorrect
else:
    print("Wrong PIN")
```

# 7. Question

A cricket academy has the following player selection rules:

    1. A player must pass the fitness test to be considered for selection.
    2. If the player passes the fitness test, then check the player's score.
    3. The player must score more than 50 runs to be selected.
    4. If the score is 50 or below, display "Score is too low for selection".
    5. If the player fails the fitness test, display "Fitness test failed".

Write a Python program using nested if-else to:

    1. Ask whether the player passed the fitness test (yes/no)
    2. Take the player's score as input only if fitness is passed
    3. Display:
        "Player Selected" if both conditions are true
        "Score is too low for selection" if fitness is passed but score is not above 50
        "Fitness test failed" if the player fails the fitness test

```python
# Asking whether player passed fitness test
fitness = input("Did the player pass fitness test? (yes/no): ")

# First condition:
# Check fitness status
if fitness == "yes":
    
    # Taking player's score input
    score = int(input("Enter player's score: "))
    
    # Second condition inside first if:
    # Check if score is above 50
    if score > 50:
        print("Player Selected")
    
    # If score is 50 or less
    else:
        print("Score is too low for selection")

# If fitness test is failed
else:
    print("Fitness test failed")
```

# 10. Find Factorial of a Number

```python
# Using While Loop
num = int(input("Enter a number: "))

fact = 1
i = 1

while i <= num:
    fact = fact * i
    i += 1

print("Factorial =", fact)
```

```python
# Write a program to count the total number of digits in a given integer using a while loop.

# Given Input: 75869

num = 75869
count = 0

# Keep dividing by 10 until the number reaches 0
while num != 0:
    # Floor division to remove the last digit
    num = num // 10
    # Increment the counter
    count = count + 1

print("Total digits are:", count)
```

```python
# 2. Count Occurrences of All Vowels
string = input("Enter a string: ").lower()

a = e = i = o = u = 0

for ch in string:
    if ch == 'a':
        a += 1
    elif ch == 'e':
        e += 1
    elif ch == 'i':
        i += 1
    elif ch == 'o':
        o += 1
    elif ch == 'u':
        u += 1

print("a =", a)
print("e =", e)
print("i =", i)
print("o =", o)
print("u =", u)
```

```python
string = input("Enter a string: ").lower()

vowels = {'a':0, 'e':0, 'i':0, 'o':0, 'u':0}

for ch in string:
    if ch in vowels:
        vowels[ch] += 1

print(vowels)
```

```python
# 1. Count Uppercase and Lowercase Characters
string = input("Enter a string: ")

upper = 0
lower = 0

for ch in string:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1

print("Uppercase:", upper)
print("Lowercase:", lower)
```

```python
# 2. Count Digits, Alphabets, and Special Characters
string = input("Enter a string: ")

digits = alphabets = special = 0

for ch in string:
    if ch.isdigit():
        digits += 1
    elif ch.isalpha():
        alphabets += 1
    else:
        special += 1

print("Digits:", digits)
print("Alphabets:", alphabets)
print("Special Characters:", special)
```

```python
# 3. Find Frequency of Each Character
string = input("Enter a string: ")

freq = {}

for ch in string:
    freq[ch] = freq.get(ch, 0) + 1

print(freq)
```

```python
# 4. Find First Non-Repeating Character
string = input("Enter a string: ")

for ch in string:
    if string.count(ch) == 1:
        print("First Non-Repeating Character:", ch)
        break
```

```python
# 5. Find First Repeated Character
string = input("Enter a string: ")

seen = set()

for ch in string:
    if ch in seen:
        print("First Repeated Character:", ch)
        break
    seen.add(ch)
```

```python
# 6. Remove Duplicate Characters

string = input("Enter a string: ")

result = ""

for ch in string:
    if ch not in result:
        result += ch

print(result)
```

```python
# 7. Count Frequency of Words
sentence = input("Enter a sentence: ")

words = sentence.split()

freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

print(freq)
```

```python
# 8. Find Largest Word
sentence = input("Enter a sentence: ")

words = sentence.split()

largest = words[0]

for word in words:
    if len(word) > len(largest):
        largest = word

print(largest)
```

```python
# 9. Find Smallest Word
sentence = input("Enter a sentence: ")

words = sentence.split()

smallest = words[0]

for word in words:
    if len(word) < len(smallest):
        smallest = word

print(smallest)
```

```python
# 10. Reverse Each Word
sentence = input("Enter a sentence: ")

words = sentence.split()

for word in words:
    print(word[::-1], end=" ")
```

```python
# 11. Reverse Word Order
sentence = input("Enter a sentence: ")

words = sentence.split()

print(" ".join(words[::-1]))
```

```python
# 15. Find Maximum Occurring Character
string = input("Enter a string: ")

max_char = ""
max_count = 0

for ch in string:
    count = string.count(ch)

    if count > max_count:
        max_count = count
        max_char = ch

print(max_char, "appears", max_count, "times")
```

```python
# 17. Remove Consecutive Duplicate Characters
string = input("Enter a string: ")

result = string[0]

for i in range(1, len(string)):
    if string[i] != string[i - 1]:
        result += string[i]

print(result)
```

```python
# 18. Compress String

string = input("Enter a string: ")

result = ""

count = 1

for i in range(len(string)-1):
    if string[i] == string[i+1]:
        count += 1
    else:
        result += string[i] + str(count)
        count = 1

result += string[-1] + str(count)

print(result)
```

# Dictionary

```python
# 1. Create a Dictionary and Print Keys and Values
student = {
    "name": "Mangesh",
    "age": 25,
    "city": "Bhopal"
}
print(student.items())
for key, value in student.items():
    print(key, ":", value)
```

```python
# 2. Count Frequency of Characters in a String
string = input("Enter a string: ")

freq = {}

for i in string: #pythoon
    freq[i] = freq.get(i, 1) + 1 #

print(freq)
```

```python
# 3. Count Frequency of Words
sentence = input("Enter a sentence: ")

words = sentence.split()
print(words)
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

print(freq)
```

```python
# 4. Find Maximum Value in Dictionary
data = {
    "A": 10,
    "B": 50,
    "C": 20
}

max_key = max(data, key=data.get)

print(max_key, data[max_key])
```

```python
# 6. Sum All Values of a Dictionary
data = {
    "A": 10,
    "B": 20,
    "C": 30
}

total = 0

for value in data.values():
    total += value

print(total)
```

```python
# 7. Merge Two Dictionaries
d1 = {"A": 1, "B": 2}
d2 = {"C": 3, "D": 4}

d1.update(d2)

print(d1)
```

```python
# 8. Check if a Key Exists
data = {
    "name": "Mangesh",
    "age": 25
}

key = input("Enter key: ")

if key in data:
    print("Key Found")
else:
    print("Key Not Found")
```

```python
# 9. Reverse a Dictionary
data = {
    "A": 1,
    "B": 2,
    "C": 3,
    "D": 1
}

reverse = {}

for key, value in data.items():
    reverse[value] = key

print(reverse)
```

```python
# 10. Sort Dictionary by Keys
data = {
    "C": 30,
    "A": 10,
    "B": 20
}

for key in sorted(data):
    print(key, data[key])
```

```python
# 11. Sort Dictionary by Values
data = {
    "A": 30,
    "B": 10,
    "C": 20
}

sorted_data = sorted(data.items(), key=lambda x: x[1])

print(sorted_data)
```

```python
# 12. Find Duplicate Values
data = {
    "A": 10,
    "B": 20,
    "C": 10,
    "D": 30
}

seen = set()

for value in data.values():
    if value in seen:
        print("Duplicate:", value)
    else:
        seen.add(value)
```

```python
# 13. Create Dictionary from Two Lists
keys = ["A", "B", "C"]
values = [10, 20, 30]

result = {}

for i in range(len(keys)):
    result[keys[i]] = values[i]

print(result)
```

```python
# 15. Group Words by Length
words = ["cat", "dog", "apple", "bat", "orange"]

result = {}

for word in words:

    length = len(word)

    if length not in result:
        result[length] = []

    result[length].append(word)

print(result)
```

```python
# 8. Find All Duplicate Elements
list1 = [1, 2, 3, 1, 2, 4, 5]
duplicates = []

for i in list1:
    if list1.count(i) > 1 and i not in duplicates:
        duplicates.append(i)

print(duplicates)
```

```python
# 9. Check Whether List is Palindrome
list1 = [1, 2, 3, 2, 1]

if list1 == list1[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 11. Product of All Elements
list1 = [1, 2, 3, 4]
product = 1

for i in list1:
    product *= i

print(product)
```

```python
# 12. Separate Even and Odd Numbers
list1 = [1, 2, 3, 4, 5, 6]

even = []
odd = []

for i in list1:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even:", even)
print("Odd:", odd)
```

```python
# 13. Move All Zeros to the End
list1 = [1, 0, 2, 0, 3, 4, 0]

result = []#1,2,3

for i in list1:
    if i != 0:
        result.append(i)

for i in list1:
    if i == 0:
        result.append(i)

print(result)
```

```python
# 14. Common Elements Between Two Lists
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

common = []

for i in list1:
    if i in list2:
        common.append(i)

print(common)
```

```python
# 14. Common Elements Between Two Lists
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

common = []

for i in range(len(list1)):
    if list1[i] in list2:
        common.append(list1[i])

print(common)
```

```python
# 15. Union of Two Lists
list1 = [1, 2, 3]
list2 = [3, 4, 5]

union = list1#[1, 2, 3]

for i in list2:
    if i not in union:
        union.append(i)

print(union)
```

```python
# 16. Intersection of Two Lists
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

intersection = []

for i in list1:
    if i in list2:
        intersection.append(i)

print(intersection)
```
