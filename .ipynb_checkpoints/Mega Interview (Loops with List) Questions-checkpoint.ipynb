```python
# 3. Find Maximum and Minimum Element in the List
numbers = [45, 12, 78, 5, 99, 34]

maximum = numbers[0]
minimum = numbers[0]

for i in numbers:
    if i > maximum:
        maximum = i

    if i < minimum:
        minimum = i

print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
# 4. Find Second Largest Number
numbers = [12, 45, 67, 89, 10]
largest = 0
second = 0  

for i in numbers:
    if i > largest:
        
        second = largest
        largest = i
        
    elif i > second and i != largest:
        second = i
        
print(second)
print(largest)
```

```python
# 15. Find All Pairs with Given Sum
numbers = [2,4,3,5,7,8,1]

target = 9

for i in range(len(numbers)): #  0-6
    for j in range(i+1,len(numbers)): #1-6
        if numbers[i] + numbers[j] == target:
            print(numbers[i], numbers[j])
```

```python
# 18. Merge Two Lists Without Duplicates
list1 = [1,2,3]
list2 = [3,4,5]

result = []

for i in list1:
    if i not in result:
        result.append(i)

for i in list2:
    if i not in result:
        result.append(i)

print(result)
```

```python
# 26. Print Prime Numbers in a List

numbers = [2,4,5,7,8,11,15]

for num in numbers:
    if num > 1:
        prime = True

        for i in range(2,num): #2-4
            if num % i == 0: #4%2
                prime = False
                break

        if prime:
            print(num)
```

```python
# 28. Fibonacci Series
n = 5

a = 0
b = 1

for i in range(n): #0-4
    print(a,end=" ") #0,1, 1, 2, 3
    c = a + b #0+1=1, 1+1=2, 1+2=3, 2+3=5, 3+5=8
    a = b #1, 1, 2, 3, 5
    b = c #1, 2, 3, 5, 8
```

```python
# 37. Move All Negative Numbers to Beginning
numbers = [5, -2, 8, -7, 10, -1, 4]

result = []

# Add negative numbers first
for i in numbers:
    if i < 0:
        result.append(i)

# Add non-negative numbers
for i in numbers:
    if i >= 0:
        result.append(i)

print(result)
```

```python
# 38. Find Second Smallest Number
numbers = [12, 5, 8, 20, 10]

smallest = numbers[0]
second = numbers[0]

# Find smallest
for i in numbers:
    if i < smallest:
        smallest = i

# Find second smallest
second = numbers[0]

for i in numbers:
    if i != smallest:
        second = i
        break

for i in numbers:
    if i < second and i != smallest:
        second = i

print("Second Smallest:", second)
```

```python
# 43. Find Pair Having Maximum Sum
numbers = [12,45,6,89,34]

maximum = numbers[0]
second = numbers[0]

for i in numbers:
    if i > maximum:
        second = maximum
        maximum = i
    elif i > second and i != maximum:
        second = i

print("Pair:", maximum, second)
print("Maximum Sum:", maximum + second)
```

```python
# 44. Find Longest String
words = ["Python","AI","Database","SQL","Programming"]

longest = words[0]

for i in words:
    if len(i) > len(longest):
        longest = i

print(longest)
```

```python
# 45. Find Shortest String
words = ["Python","AI","Database","SQL","Programming"]

shortest = words[0]

for i in words:
    if len(i) < len(shortest):
        shortest = i

print(shortest)
```

```python
# 46. Count Vowels in Every String
words = ["Python","Apple","SQL","Education"]

vowels = "aeiouAEIOU"

for word in words:

    count = 0

    for letter in word:

        if letter in vowels:
            count += 1

    print(word, ":", count)
```

```python
# 48. Remove Consecutive Duplicate Elements
numbers = [1,1,2,2,3,3,4,5,5,4]

result = []

for i in numbers:

    if len(result) == 0 or result[-1] != i:
        result.append(i)

print(result)
```
