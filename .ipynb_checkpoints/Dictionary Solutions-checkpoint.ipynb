```python
# 9. Count the Frequency of Each Character in a String

text = input("Enter a string: ")

freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)
```

```python
# 10. Count the Frequency of Each Word in a Sentence

sentence = input("Enter a sentence: ")

words = sentence.split()
print(words)

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```

```python
# 11. Find the First Non-Repeating Character in a String
text = input("Enter a string: ")

freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
print(freq)
for ch in text:
    if freq[ch] == 1:
        print("First Non-Repeating Character:", ch)
        break
```

```python
# 17. Find the Key Having the Maximum Value
d = {"A": 10, "B": 50, "C": 30}

max_key = max(d, key=d.get)

print("Key =", max_key)
```

```python
# Print all the keys and values from a dictionary
student = {
    "Name": "Mangesh",
    "Age": 22,
    "Marks": 92
}

print("Keys and Values:")

for key, value in student.items():
    print(key, ":", value)
```

```python
# Find and print the key(s) corresponding to a given value
student = {
    "Name": "Mangesh",
    "Age": 22,
    "Marks": 92,
    "Score": 92
}

search_value = 92

print("Keys having value", search_value)

for key, value in student.items():
    if value == search_value:
        print(key)
```

```python
# Remove a key-value pair using a given key
student = {
    "Name": "Mangesh",
    "Age": 22,
    "Marks": 92
}

key = "Age"

if key in student:
    del student[key]

print(student)
```

```python
# Find the length of a dictionary without using len()
student = {
    "Name": "Mangesh",
    "Age": 22,
    "Marks": 92
}

count = 0

for i in student:
#     print(i)
    count += 1

print("Length =", count)
```

```python
# Invert a dictionary (handle duplicate values)
student = {
    "A": 90,
    "B": 85,
    "C": 90,
    "D": 80
}

inverse = {}

for key, value in student.items():
    if value not in inverse:
        inverse[value] = [key]
    else:
        inverse[value].append(key)

print(inverse)
```

```python
# Merge two dictionaries and sum common values
dict1 = {
    "A": 10,
    "B": 20,
    "C": 30
}

dict2 = {
    "B": 15,
    "C": 25,
    "D": 40
}

result = dict1.copy()

for key, value in dict2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print(result)
```

```python
# Count the frequency of each word in a paragraph
paragraph = """
Python is easy to learn.
Python is powerful.
Python is popular.
"""

words = paragraph.lower().replace(".", "").split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
# Filter a dictionary based on a threshold
marks = {
    "Rahul": 45,
    "Mangesh": 92,
    "Amit": 78,
    "Neha": 35
}

threshold = 50

filtered = {}

for key, value in marks.items():
    if value > threshold:
        filtered[key] = value

print(filtered)
```
