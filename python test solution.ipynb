```python
# 1. Check Whether a Number Lies Between 10 and 100 and Is Divisible by Both 5 and 11
num = int(input("Enter a number: "))

if 10 <= num <= 100 and num % 5 == 0 and num % 11 == 0:
    print("Condition Satisfied")
else:
    print("Condition Not Satisfied")
```

```python
# 7. Find the Greatest Digit in a Number
num = int(input("Enter a number: "))

greatest = 0

while num > 0:
    digit = num % 10

    if digit > greatest:
        greatest = digit

    num = num // 10

print("Greatest Digit =", greatest)
```

```python
# 10. ATM Withdrawal System

correct_pin = 1234
balance = 10000

pin = int(input("Enter PIN: "))

if pin == correct_pin:

    amount = float(input("Enter withdrawal amount: "))

    if amount <= balance:
        balance = balance - amount

        print("Withdrawal Successful")
        print("Remaining Balance =", balance)

    else:
        print("Insufficient Balance")

else:
    print("Wrong PIN")
```
