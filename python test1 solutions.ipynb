```python
# Q11. Reverse a Number
num = int(input("Enter Number: "))

reverse = 0

while num > 0:

    digit = num % 10

    reverse = reverse * 10 + digit

    num = num // 10

print("Reversed Number:", reverse)
```

```python
# Q12. Palindrome Number

num = 121
original = num
reverse = 0

while num > 0:

    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Q13. Armstrong Number
num = 153

temp = num
sum1 = 0

while num > 0:

    digit = num % 10

    sum1 = sum1 + (digit ** 3)

    num = num // 10

if temp == sum1:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
# Q15. Greatest Digit in a Number
num=9167
num = int(input("Enter Number: "))

greatest = 0 #9

while num > 0:

    digit = num % 10

    if digit > greatest:
        greatest = digit

    num = num // 10

print("Greatest Digit:", greatest)
```

```python
# Q18. Prime Number
num = int(input("Enter a number: "))
11
if num > 1:
    # Check for factors from 2 to num-1
    for i in range(2, num):
        if (num % i) == 0:
            print(f"{num} is not a prime number")
            break
    else:
        # If the loop finishes without finding a factor
        print(f"{num} is a prime number")
else:
    # Numbers less than or equal to 1 are not prime
    print(f"{num} is not a prime number")
```

```python
# ATM Withdrawal Program

correct_pin = 1234
balance = 10000

pin = int(input("Enter PIN: "))

if pin == correct_pin:
    amount = float(input("Enter withdrawal amount: "))

    if amount <= balance:
        balance -= amount
        print("Withdrawal Successful")
        print("Remaining Balance:", balance)
    else:
        print("Insufficient Balance")
else:
    print("Wrong PIN")
```
